```python
import pandas as pd

weather_df = pd.read_csv("weather_cleaned.csv")
geographic_df = pd.read_csv("geographic_df.csv")
time_df = pd.read_csv("TimeSeverityColumns.csv")
car_data = pd.read_csv("US_Accidents_March23_sampled_500k.csv")

print(f"Weather: {weather_df.shape}")
print(f"Geographic: {geographic_df.shape}")
print(f"Time: {time_df.shape}")
print(f"Original: {car_data.shape}")
```

```python
# We found that Infrastructure didn't require any cleaning 
# Select infrastructure and road-related variables
infrastructure_columns = [
    "Bump",
    "Crossing",
    "Give_Way",
    "Junction",
    "Railway",
    "Roundabout",
    "Stop",
    "Traffic_Signal",
    "Sunrise_Sunset"
]

infrastructure_df = car_data[infrastructure_columns].copy()
infrastructure_df.head()
```

```python
# Combine weather, time, geographic, and infrastructure data
combined_df = pd.concat(
    [
        weather_df.reset_index(drop=True),
        geographic_df.reset_index(drop=True),
        infrastructure_df.reset_index(drop=True)
    ],
    axis=1
)

print("Combined dataset:", combined_df.shape)
combined_df.head()
```

```python
# Add Time Data new variables
time_columns = [
    "ID", "End_Time", "Distance(mi)",
    "Description", "Start_Hour", "End_Hour"
]

final_df = combined_df.merge(
    time_df[time_columns],
    on="ID",
    how="left",
    validate="one_to_one"
)

print("Final dataset:", final_df.shape)
```

```python
final_df.to_csv("combined_cleaned.csv", index=False) # Save the final cleaned dataset to a CSV file
```

Code for cleaning dataset columns ['ID', 'Start_Time', 'End_Time', 'Distance(mi)', 'Severity', 'Description']

```python
# Import python libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data from CSV file
df = pd.read_csv('US_Accidents_March23_sampled_500k.csv')
```

Initial Profiling and Exploration

```python
# Discover shape and info of selected columns
df_cleaned = df[['ID', 'Severity', 'Start_Time', 'End_Time', 'Distance(mi)', 'Description']].copy()

print(df_cleaned.shape)
df_cleaned.info()
df_cleaned.describe()
```

```python
# Fill in null values in Description
df_cleaned['Description'] = df_cleaned['Description'].fillna('No description')
print("Missing Descriptions:", df_cleaned['Description'].isna().sum())
df_cleaned.info()
```

```python
# Determine outliers in Distance(mi) using IQR

# Log transformation
df_cleaned['Distance_log'] = np.log1p(df_cleaned['Distance(mi)'])

# Calculate Q1 and Q3
Q1 = df_cleaned['Distance_log'].quantile(0.25)
Q3 = df_cleaned['Distance_log'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# Identify outliers
outliers = (
    (df_cleaned['Distance_log'] < lower_bound) |
    (df_cleaned['Distance_log'] > upper_bound)
)

# Set outlier distances to null
df_cleaned.loc[outliers, 'Distance(mi)'] = np.nan

# Remove temporary log column
df_cleaned.drop(columns=['Distance_log'], inplace=True)
```

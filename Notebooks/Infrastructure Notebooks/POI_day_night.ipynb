```python
#https://drive.google.com/drive/folders/1AiRxKsu2j7hqARbPqlsJJlCGlF28Kl3K?usp=drive_link
df = pd.read_csv("/combined_cleaned.csv")
cols = ["Bump", "Crossing", "Give_Way", "Junction", "Railway",
        "Roundabout", "Stop", "Traffic_Signal"]
night_day = "Sunrise_Sunset"
```

```python
# titanic.groupby(["Sex", "Pclass"])["Fare"].mean()
results = {}
for col in cols:
    results[col] = df.groupby([col, "Sunrise_Sunset"])["Severity"].mean()
means = pd.DataFrame(results).T
```

```python
color, presence = "red", ""
for(i, j) in means.columns:
    if i:
        presence = "present"
        color = "green"
    else:
        presence = "absent"
        color = "#bf3636"

    if j == "Day":
        marker = "$\u2600$"
    else:
        marker = "$\u263E$"

    label = f"POI {presence}, {j}"
    plt.scatter(means[i, j], means.index, color=color, label=label, marker=marker, s=100)
plt.xlabel("Average Severity")
plt.legend()
plt.show()
```

```python
difference = (means[(False, "Day")] - means[(True, "Day")]).sort_values()
colors = []
for i in difference:
    if i < 0:
        colors.append("green")
    else:
        colors.append("#bf3636")
plt.barh(difference.index, difference, color = colors)
plt.show
```

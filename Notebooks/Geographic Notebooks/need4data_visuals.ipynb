```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/combined_cleaned.csv')
```

```python
sns.set_style(style= 'darkgrid')
```

```python
ten_city = df['City'].value_counts().head(10)
```

```python
ten_street = df['Street'].value_counts().head(10)
```

```python
ten_county = df['County'].value_counts().head(10)
```

```python
ten_zipcode = df['Zipcode'].value_counts().head(10)
```

```python
sns.set_style(style= 'darkgrid')
sns.barplot(x=df['County'].value_counts().head(10).index, y=ten_county.values, color= '#BF3636')
sns.despine()

plt.ylabel('Number of Accidents', fontsize=12, color= 'White')
plt.xlabel('County', fontsize=12, color= 'White')
plt.xticks(rotation=90)
plt.xticks(fontsize=10, color= 'White', )
plt.yticks(color= 'White')
plt.gcf().patch.set_facecolor('Black')
plt.show()
```

```python
sns.barplot(x=df['City'].value_counts().head(10).index, y=ten_city.values, color= '#BF3636')
sns.despine()

plt.ylabel('Number of Accidents', fontsize=12, color= 'White')
plt.xlabel('City', fontsize=12, color= 'White')
plt.xticks(rotation=90)
plt.xticks(fontsize=10, color= 'White', )
plt.yticks(color= 'White')
plt.gcf().patch.set_facecolor('Black')
plt.show()
```

```python
sns.set_style(style= 'darkgrid')
sns.barplot(x=df['Street'].value_counts().head(10).index, y=ten_street.values, color= '#BF3636')
sns.despine()

plt.ylabel('Number of Accidents', fontsize=12,color= 'White')
plt.xlabel('Street', fontsize=12, color= 'White')
plt.xticks(rotation=90, color= 'White')
plt.xticks(fontsize=10)
plt.yticks(color= 'White')
plt.gcf().patch.set_facecolor('Black')

plt.show()
```

```python
sns.set_style(style= 'darkgrid')
sns.barplot(x=df['Zipcode'].value_counts().head(10).index, y=ten_zipcode.values, color= '#BF3636')
sns.despine()

plt.ylabel('Number of Accidents', fontsize=12, color= 'White')
plt.xlabel('Zipcode', fontsize=8, color= 'White')
plt.xticks(rotation=90)
plt.xticks(fontsize=10, color= 'White', )
plt.yticks(color= 'White')
plt.gcf().patch.set_facecolor('Black')
plt.show()
```
